**c. First shots at visualizing the data**

Take a look at the data as plots to get more insight to the patterns, trends, and relationships. This could be histograms, scatter plots, box plots, ect. What did you gather? What stands out?

```python
import seaborn as sns

#ride prices distribution
plt.hist(df["price"].dropna(), bins = 30, edgecolor = "black")

plt.title("Distribution of ride prices")
plt.xlabel("Price")
plt.ylabel("Nr of rides")
plt.show()

#average price by servide type (lux, xl, shared etc)
avg_prices = (
    df.groupby("name")["price"]
    .mean()
    .dropna()
    .sort_values()
)

avg_prices.plot(kind = "barh")

plt.title("Average price by service type")
plt.xlabel("Average price")
plt.ylabel("Service type")
plt.show()

#heatmap keskmine hind iga pickup-dropoff vahel

heatmap_data = df.pivot_table(
    values = "price",
    index = "source",
    columns = "destination",
    aggfunc = "mean"
)

sns.heatmap(
    heatmap_data,
    annot = True, #näitab kastides hinda
    fmt = ".1f", #näitab 1 komakohaga
    cmap = "magma",
)

plt.title("Average prices by each pickup and dropoff")
plt.xlabel("Dropoff")
plt.ylabel("Pickup")

plt.show()
```

**Answer:** On the first graph we can see that the distribution is very right-skewed. Most rides cost less than 20$ and rides that cost more than 40 dollars are pretty rare. The second graph shows the average price of the different service types there are. Shared rides are the cheapest whilst luxury XL cars are the most expensive. This shows how big of an impact the service type has on price. The heatmap shows average prices between different pickup and drop-off locations. Some routes are more expensive than others. This may be because of distance but also what type of service is used more often in which route. The shape of the heatmap also stands out - it's a very clean pattern which makes us believe the data is at the very least heavily sampled but most likely synthesized.

**d. Check for data quality**

Missing values, duplicates, impossible or inconsistent values. Also watch out for outliers and anomalies that may require further investigation. Explain what you found and what you are going to do with the information.


```python
# missing values
print("Missing values:")
print(df.isnull().sum())

#duplicate rows
print("\nDuplicate rows:")
print(df.duplicated().sum())

#impossible values in prices and distance
print("\nNegative or zero prices:")
print((df["price"] <= 0).sum())

print("\nNegative or zero distances:")
print((df["distance"] <= 0).sum())

# price outliers
print("\nPrice statistics:")
print(df["price"].describe())

# unusually high prices
print("\nPrices above $60:")
print((df["price"] > 60).sum())
```

**Answer:** The main issue would be the 55095 rows, where the prices are missing. We plan to remove these rows before building the prediction model as they can't be used to train the model or later test it, because we won't know whether our model got the price correct or missed by a lot. There are no duplicate rows and we also didn't find any negative/zero prices or distances. We found that there are 590 rows, where the price is over 60$ which seems quite excessive, but can also be valid prices for longer rides with luxury services. Also whilst just looking at the table we noticed, that visibility and visibility.1 are the same so one of these can be removed. Overall the dataset is suitable for our project, but does need some minor cleaning.
